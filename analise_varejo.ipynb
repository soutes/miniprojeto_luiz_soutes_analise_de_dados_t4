## Sprint 1 (Importação dos dados)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Criação do dataframe
df = pd.read_csv("data/base_varejo.csv", sep=";")
```

```python
df.info()
```

Dataframe criado. São 83 mil registros distribuidos em 14 colunas. 4 Colunas estão com nome unnamed e precisão ser exploradas com cuidado.

### Dicionário de dados:
1. DATA: Data da compra;

2. CO_ID: Identificação do número de compra (número da nota fiscal);

3. CL_ID: Identificação do cliente (número do cliente);

4. CL_GENERO: Sexo biológico informado pelo cliente;

5. CL_EC: Estado civil do cliente:
    1: Casado ou união estával;
    2: Divorciado;
    3: Separado;
    4: Solteiro;
    5: Viúvo.

6. CL_FHL: Número de filhos do cliente;

7. CL_SEG: Segmentação econômica do cliente (classe A, B ou C);

8. PR_ID: Código do produto (SKU) adquirido;

9. PR_CAT: Categoria do produto adquirido;

10. PR_NOME: Nome do produto adquirido.

## Sprint 2 (Transformação de Strings, Integer e Float e Datetime):

```python

#Conversão da DATA para datetime
df['DATA'] = pd.to_datetime(df['DATA'], format='%d/%m/%Y')
```

```python
# Conversão da colunas IDs para o tipo string (colunas com ID nao tem função matemática e podem acabar gerando viés na análise)
df['CO_ID'] = df['CO_ID'].astype(str)
df['CL_ID'] = df['CL_ID'].astype(str)
df['PR_ID'] = df['PR_ID'].astype(str)
df['CL_EC'] = df['CL_EC'].astype(str)
```

# Exploração dos dados

```python
vendas_por_dia = df.groupby('DATA').size()
print(vendas_por_dia)
```

```python
print(vendas_por_dia.min())
print(vendas_por_dia.max())
```

A quantidade de vendas varia de 403 a 4641.

```python
plt.figure(figsize=(12, 5))
plt.plot(vendas_por_dia.index, vendas_por_dia.values)
plt.title('Vendas por Dia')
plt.xlabel('Data')
plt.ylabel('Quantidade')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

Pode-se observar uma variabilidade grande no número de vendas.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['CL_FHL'])
plt.title('Boxplot Número de Filhos (CL_FHL)', fontsize=14)
plt.ylabel('CL_FHL')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

Com o boxplot podemos verificar que não existem outliers na coluna númerica (número de filhos) sendo que o maior número é 4, o menor é 0 e a maioria dos registros estão entre 0 e 2.

```python
print(df['CL_FHL'].describe())
```

Analisando as estatísticas, o número de filhos dos clientes varia de 0 a 4, como vimos no boxplot. A média é de 1,14 filho por cliente, mas a mediana é 0, ou seja, a maioria dos clientes não tem filhos. Além disso, 75% dos clientes têm até 2 filhos.
